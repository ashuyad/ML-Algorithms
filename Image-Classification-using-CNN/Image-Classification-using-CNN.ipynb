```python
import tensorflow as tf
from tensorflow.keras import layers , models
```

```python
data_augmentation = models.Sequential([
    layers.Rescaling(1./255),
    layers.RandomShear(x_factor=0.2, y_factor=0.2),
    layers.RandomZoom(height_factor = 0.2, width_factor = 0.2),
    layers.RandomFlip('horizontal')
])

training_set = tf.keras.utils.image_dataset_from_directory('dataset/training_set', image_size= (64, 64), batch_size = 32, label_mode ='binary')

my_saved_classes = training_set.class_names
training_set = training_set.map(
    lambda x, y: (data_augmentation(x, training=True), y),
    num_parallel_calls=tf.data.AUTOTUNE)

training_set = training_set.prefetch(buffer_size=tf.data.AUTOTUNE)
```

```python
data_augmentation = models.Sequential([
    layers.Rescaling(1./255)
])

test_set = tf.keras.utils.image_dataset_from_directory('dataset/test_set', image_size= (64, 64), batch_size = 32, label_mode ='binary')

test_set = test_set.map(
    lambda x, y: (data_augmentation(x, training=False), y),
    num_parallel_calls=tf.data.AUTOTUNE)

training_set = training_set.prefetch(buffer_size=tf.data.AUTOTUNE)
```

```python
cnn = tf.keras.models.Sequential()
```

```python
cnn.add(tf.keras.layers.Conv2D(filters= 32, activation = 'relu', kernel_size = 3, input_shape=[64, 64, 3]))
```

```python
cnn.add(tf.keras.layers.MaxPool2D(pool_size = 2, strides = 2))
```

```python
cnn.add(tf.keras.layers.Conv2D(filters= 32, activation = 'relu', kernel_size = 3))
cnn.add(tf.keras.layers.MaxPool2D(pool_size = 2, strides = 2))
```

```python
cnn.add(tf.keras.layers.Flatten())
```

```python
cnn.add(tf.keras.layers.Dense(units=128, activation='relu'))
```

```python
cnn.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))
```

```python
cnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
cnn.fit(x = training_set, validation_data= test_set, epochs= 25)
```

```python
import numpy as np
from keras.preprocessing import image
test_image = image.load_img('dataset/single_prediction/cat_or_dog.jpg', target_size = (64, 64))
test_image = image.img_to_array(test_image)
test_image = np.expand_dims(test_image, axis=0)
result = cnn.predict(test_image)

if result[0][0]==1:
    prediction = 'dog'
else:
    prediction = 'cat'
```
